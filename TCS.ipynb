```python
from pyspark.sql.functions import col, floor, max, concat, lit

def result_1(input_df):
    # 1. Convert price to numeric (integer/double) first, as CSV reads are strings
    # Then calculate INR: floor(price * 86.61)
    df = input_df.withColumn("price_inr", floor(col("price").cast("double") * 86.61))

    # 2. Find the maximum price based on each "neighbourhood_group"
    # We group by neighbourhood_group and aggregate the max of our new price_inr column
    #df_max = df_inr.groupby("neighbourhood_group").agg(max("price_inr").alias("max_val"))
    df = df.groupBy("neighbourhood_group").agg(max("price_inr").alias("max_price_inr"))

    df = df.withColumn("max_price_inr", concat(lit("INR"), col("max_price_inr").cast("int").cast("string")))

    # 3. Add the string "INR" at the foremost of each value
    # We verify the column name matches the sample output: "max_price_inr"

    # Select only the required columns in the correct order
    return df.select("neighbourhood_group", "max_price_inr")
```

```python
from pyspark.sql.functions import col, count, sum, desc

def result_2(input_df):
    # 1. Filter rows: review_rate_number >= 4 AND host_identity_verified == 'verified'
    # We cast review_rate_number to integer to ensure correct comparison
    filtered_df = input_df.filter(
        (col("review_rate_number").cast("int") >= 4) &
        (col("host_identity_verified") == "verified")
    )

    # 2. Group by "host_name" and calculate aggregates
    # We sum (price + service_fee). We cast them to double to ensure arithmetic works.
    grouped_df = filtered_df.groupBy("host_name").agg(
        count("host_name").alias("count_of_hotels"),
        sum(col("price").cast("double") + col("service_fee").cast("double")).alias("sum_of_total_price")
    )

    # 3. Filter for hosts with 5 or more hotels
    final_filtered_df = grouped_df.filter(col("count_of_hotels") >= 5)

    # 4. Sort by sum_of_total_price in descending order
    sorted_df = final_filtered_df.orderBy(desc("sum_of_total_price"))

    # 5. Return the final dataframe
    return sorted_df
```

```python
from pyspark.sql.functions import col, round, when, count, dense_rank
from pyspark.sql.window import Window

def result_3(input_df):
    # 1. Create "fee_percentage"
    # Formula: (service_fee / price) * 100
    # Cast columns to double to ensure division works correctly
    df_pct = input_df.withColumn("fee_percentage",
        round((col("service_fee").cast("double") / col("price").cast("double")) * 100, 2)
    )

    # 2. Create "fee_category" using conditional logic (when/otherwise)
    df_cat = df_pct.withColumn("fee_category",
        when(col("fee_percentage") < 10, "Low Fee")
        .when((col("fee_percentage") >= 10) & (col("fee_percentage") <= 20), "Moderate Fee")
        .otherwise("High Fee")
    )

    # 3. Aggregation: Count of "id" named "listing_count"
    # Group by Construction_year, room_type, and the new fee_category
    df_grouped = df_cat.groupBy("Construction_year", "room_type", "fee_category").agg(count("id").alias("listing_count"))

    # 4. Ranking: "dense_rank" (no holes) partitioned by Year, ordered by count DESC
    windowSpec = Window.partitionBy("Construction_year").orderBy(col("listing_count").desc())

    final_df = df_grouped.withColumn("rank", dense_rank().over(windowSpec))

    # 5. Select the final columns in the specific order required
    return final_df.select("Construction_year", "room_type", "fee_category", "listing_count", "rank")
```

```python
from sklearn.preprocessing import LabelEncoder
# 1. Satisfy the requirement: Create the object and store it in 'le_object'
le_object = LabelEncoder()

# 2. Implied Task: actually APPLY it to the data
# (If you don't do this loop, the model training in the next step will fail)
for col in data.columns:
    if data[col].dtype == 'object':
        data[col] = le_object.fit_transform(data[col])

print(data)
```

```python
# Store independent variables (Drop 'price' column)
X = data.drop(columns=['price'])

# Store dependent variable (Select only 'price' column)
y = data['price']
```

```python
from sklearn.model_selection import train_test_split

# Perform the split according to instructions
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=20)
```

```python
from sklearn.linear_model import LinearRegression

# 1. Store the model in variable 'model'
model = LinearRegression()

# 2. Train the model (Use .fit with the training data)
model.fit(X_train, y_train)

# 3. Predict values (Use .predict with the test data)
predictions = model.predict(X_test)

predictions
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# 1. Mean Absolute Error (MAE)
mae = mean_absolute_error(y_test, predictions)

# 2. Mean Squared Error (MSE)
mse = mean_squared_error(y_test, predictions)

# 3. Root Mean Squared Error (RMSE)
# (Numpy sqrt of MSE)
rmse = np.sqrt(mse)

# 4. R2 Score (Accuracy/Fit)
r2 = r2_score(y_test, predictions)

# Print all measures
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```
